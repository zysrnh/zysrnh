# MODUL 2: Preprocessing Teks Bisnis (Pembersihan Data Teks) dengan NLTK & Sastrawi
**Tujuan Pembelajaran:** Menyiapkan teks ulasan atau komplain pelanggan yang mentah dan tidak terstruktur agar bersih, konsisten, dan siap diolah oleh model AI.


### Modul 2 - Langkah 2.1: Case Folding (Lowercase) & Pembersihan Simbol Teks via Regex
Langkah pertama adalah menyeragamkan semua huruf menjadi huruf kecil (*case folding*) dan membersihkan tanda baca, angka, simbol emotikon, serta spasi ganda menggunakan modul `re` (Regular Expression).


```python
# ==============================================================
# MODUL 2 - LANGKAH 2.1: CASE FOLDING & REGEX CLEANING
# ==============================================================

# Teks ulasan komplain mentah dari pelanggan
teks_mentah = "Pengiriman Kurir SANGAT Lambat!! Paket Saya Rusak & Terlambat Sampai di Alamat :-("

# 1. Case Folding: Ubah semua huruf menjadi lowercase
teks_lower = teks_mentah.lower()

# 2. Regex Cleaning: Hapus tanda baca, simbol emotikon, dan angka, sisakan huruf saja
teks_bersih = re.sub(r'[^a-zA-Z\s]', ' ', teks_lower)

# 3. Hapus spasi berlebih
teks_bersih = re.sub(r'\s+', ' ', teks_bersih).strip()

print(f"Teks Mentah : {teks_mentah}")
print(f"Teks Bersih : {teks_bersih}")
```

### Modul 2 - Langkah 2.2: Tokenisasi Teks (NLTK `word_tokenize`)
Langkah kedua adalah memecah kalimat teks yang sudah bersih menjadi potongan kata-kata tunggal (*tokens*) menggunakan fungsi `word_tokenize` dari pustaka NLTK.


```python
# ==============================================================
# MODUL 2 - LANGKAH 2.2: TOKENISASI KATA MENGGUNAKAN NLTK
# ==============================================================

# Fungsi tokenisasi dengan fallback jika punkt belum terpasang
def tokenisasi_kalimat(teks):
    if 'nltk' in globals() and nltk is not None:
        try:
            return nltk.word_tokenize(teks)
        except Exception:
            pass
    # Fallback pemotongan berbasis spasi jika resource NLTK tidak tersedia
    return teks.split()

tokens_ulasan = tokenisasi_kalimat(teks_bersih)
print(f"Hasil Tokenisasi ({len(tokens_ulasan)} kata):")
print(tokens_ulasan)
```

### Modul 2 - Langkah 2.3: Stopword Removal & Stemming Bahasa Indonesia
Langkah ketiga adalah menghapus kata-kata penghubung umum yang tidak bermakna klasifikasi (*stopword removal* seperti `di`, `dan`, `saya`), kemudian mengembalikan kata berimbuhan ke bentuk dasarnya (*stemming*, misalnya `pengiriman` → `kirim`, `terlambat` → `lambat`).


```python
try:
    import Sastrawi
except ImportError:
    import sys, subprocess
    subprocess.check_call([sys.executable, "-m", "pip", "install", "Sastrawi", "-q"])
# ==============================================================
# MODUL 2 - LANGKAH 2.3: STOPWORDS REMOVAL & STEMMING SASTRAWI
# ==============================================================

# Kamus stopword standar bahasa Indonesia
daftar_stopwords = {"dan", "di", "ke", "dari", "saya", "ini", "itu", "yang", "pada", "untuk", "dengan"}

# 1. Stopword Removal: Saring kata-kata yang bukan stopword
tokens_tanpa_stopword = [kata for kata in tokens_ulasan if kata not in daftar_stopwords]

# 2. Stemming Sederhana (Menggunakan Sastrawi jika tersedia, atau kamus aturan dasar)
def stem_kata_indonesia(kata_list):
    try:
        from Sastrawi.Stemmer.StemmerFactory import StemmerFactory
        factory = StemmerFactory()
        stemmer = factory.create_stemmer()
        return [stemmer.stem(k) for k in kata_list]
    except ImportError:
        # Aturan stemming dasar untuk kata berimbuhan umum jika Sastrawi belum diinstal
        kamus_dasar = {
            "pengiriman": "kirim", "terlambat": "lambat", "sampai": "sampai",
            "rusak": "rusak", "kurir": "kurir", "sangat": "sangat", "alamat": "alamat"
        }
        return [kamus_dasar.get(k, k) for k in kata_list]

tokens_stemmed = stem_kata_indonesia(tokens_tanpa_stopword)
teks_hasil_preprocessing = " ".join(tokens_stemmed)

print(f"Tokens Setelah Stopword Removal : {tokens_tanpa_stopword}")
print(f"Tokens Setelah Stemming         : {tokens_stemmed}")
print(f'Hasil Akhir Teks Preprocessed   : "{teks_hasil_preprocessing}"')
```

# MODUL 3: Representasi Teks & Ekstraksi Fitur (Bag of Words & TF-IDF)
**Tujuan Pembelajaran:** Mengonversi data teks pelanggan yang sudah bersih menjadi matriks angka numerik (vektor fitur) agar dapat dianalisis dan diproses oleh algoritma Machine Learning.


### Modul 3 - Langkah 3.1: Vektorisasi Frekuensi Kata (*Bag of Words* / CountVectorizer)
*Bag of Words* (BoW) menyusun kamus kosakata unik (*vocabulary*) dari seluruh dokumen, lalu menghitung berapa kali setiap kata muncul pada masing-masing kalimat ulasan pelanggan.


```python
# ==============================================================
# MODUL 3 - LANGKAH 3.1: BAG OF WORDS MENGGUNAKAN COUNTVECTORIZER
# ==============================================================

# Contoh 4 kalimat pesan masuk dari pelanggan bisnis
corpus_bisnis = [
    "halo admin mau tanya tarif ongkir ke surabaya",
    "tolong cek lokasi paket pengiriman resi saya",
    "paket saya rusak dan pengiriman kurir lambat",
    "berapa biaya tarif kirim paket ke kota bandung"
]

# Inisialisasi CountVectorizer dari Scikit-Learn
bow_vectorizer = CountVectorizer()
X_bow = bow_vectorizer.fit_transform(corpus_bisnis)

# Tampilkan kosakata unik yang dipelajari sistem
vocab = bow_vectorizer.get_feature_names_out()
print("Daftar Kosakata Unik (Vocabulary):")
print(vocab)
print(f"\nDimensi Matriks BoW: {X_bow.shape} (4 Dokumen x {len(vocab)} Kata Unik)")
print("\nMatriks Frekuensi Kata (Array):")
print(X_bow.toarray())
```

### Modul 3 - Langkah 3.2: Pembobotan Nilai Kepentingan Kata (*TF-IDF*)
*TF-IDF* (Term Frequency - Inverse Document Frequency) memberi nilai bobot tinggi pada kata-kata penting yang spesifik untuk topik tertentu (seperti `rusak` atau `ongkir`), dan menurunkan bobot kata yang muncul terlalu sering di semua dokumen.


```python
# ==============================================================
# MODUL 3 - LANGKAH 3.2: PEMBOBOTAN TF-IDF (TFIDFVECTORIZER)
# ==============================================================

# Inisialisasi TfidfVectorizer
tfidf_vectorizer = TfidfVectorizer()
X_tfidf = tfidf_vectorizer.fit_transform(corpus_bisnis)

print("Matriks Bobot TF-IDF untuk Dokumen ke-3 ('paket saya rusak dan pengiriman kurir lambat'):")
fitur = tfidf_vectorizer.get_feature_names_out()
bobot_dokumen_3 = X_tfidf.toarray()[2]

for idx, kata in enumerate(fitur):
    if bobot_dokumen_3[idx] > 0:
        print(f" - Kata: '{kata:<12}' ↳ Bobot TF-IDF: {bobot_dokumen_3[idx]:.4f}")
```

### Modul 4 - Langkah 4.1: Merancang Kamus Intent Berformat JSON / Dictionary
Struktur kamus intent memuat tag kategori (misalnya `sapaan`, `tanya_tarif`, `lacak_resi`, `komplain`), daftar pola pertanyaan yang sering ditanyakan user, serta variasi template jawaban bisnis.


```python
# ==============================================================
# MODUL 4 - LANGKAH 4.1: DESAIN KAMUS INTENT BISNIS (DICTIONARY)
# ==============================================================

kamus_intent_fastsend = {
    "intents": [
        {
            "tag": "sapaan",
            "patterns": ["halo admin", "selamat pagi", "hai fastsend", "permisi kak"],
            "responses": [
                "Halo! Selamat datang di Layanan Pelanggan FastSend Express. Ada yang bisa kami bantu?",
                "Hai Kak! Senang bisa melayani Anda di FastSend Express. Silakan sampaikan pertanyaan Anda."
            ]
        },
        {
            "tag": "tanya_tarif",
            "patterns": ["berapa tarif ongkir", "biaya kirim ke surabaya", "cek harga ongkos kirim"],
            "responses": [
                "Untuk cek tarif ongkir resmi, silakan sebutkan kota asal, kota tujuan, dan perkiraan berat paket (kg).",
                "Tarif reguler mulai dari Rp 10.000/kg. Anda juga bisa mengecek kalkulator tarif lengkap di web FastSend."
            ]
        },
        {
            "tag": "lacak_resi",
            "patterns": ["tolong cek resi", "lacak keberadaan paket", "posisi kiriman saya dimana"],
            "responses": [
                "Silakan sertakan nomor resi pengiriman Anda (contoh: resi: 99482101) agar dapat kami lacak secara otomatis.",
                "Sistem kami siap melacak paket Anda. Mohon ketik nomor resi pengiriman Anda."
            ]
        },
        {
            "tag": "komplain",
            "patterns": ["paket saya rusak", "kurir sangat lambat", "kiriman barang belum sampai"],
            "responses": [
                "Mohon maaf atas ketidaknyamanan Anda. Tim kendali mutu kami akan segera menginvestigasi kendala pengiriman ini.",
                "Kami memohon maaf atas kendala yang dialami. Silakan laporkan nomor resi dan foto paket Anda untuk klaim garansi."
            ]
        }
    ]
}

print("Kamus Intent Berhasil Dibuat!")
print(f"Total Kategori Intent: {len(kamus_intent_fastsend['intents'])} Kategori")
for item in kamus_intent_fastsend["intents"]:
    print(f" * Tag: [{item['tag']:<12}] -> Memiliki {len(item['patterns'])} contoh pola pertanyaan.")
```

### Modul 4 - Langkah 4.2: Logika Pattern Matching & Respon Otomatis Sederhana
Fungsi berikut mencocokkan kata kunci dari pesan pelanggan terhadap pola yang ada pada kamus intent, lalu memilih respon jawaban secara acak menggunakan `random.choice`.


```python
# ==============================================================
# MODUL 4 - LANGKAH 4.2: PENCOCOKAN POLA & PEMILIHAN RESPON
# ==============================================================

def jawab_pesan_sederhana(pesan_masukan):
    pesan_lower = pesan_masukan.lower()
    
    for item in kamus_intent_fastsend["intents"]:
        for pola in item["patterns"]:
            # Jika kata dalam pola ditemukan di dalam pesan masukan
            if any(kata in pesan_lower for kata in pola.split()):
                jawaban = random.choice(item["responses"])
                return item["tag"], jawaban
                
    return "unknown", "Mohon maaf, kami belum dapat memahami pesan Anda. Silakan hubungi CS kami."

# Simulasi pengujian pesan
tag_hasil, respon_hasil = jawab_pesan_sederhana("halo admin fastsend, selamat pagi")
print('Pesan Pengguna   : "halo admin fastsend, selamat pagi"')
print(f"Intent Terdeteksi: [{tag_hasil}]")
print(f'Respon Chatbot   : "{respon_hasil}"')
```

# MODUL 5: Rule-Based Chatbot & Ekstraksi Entitas Regex
**Tujuan Pembelajaran:** Menggunakan Regular Expression (`re.search`) untuk mengekstrak entitas penting (seperti nomor resi atau nomor telepon) secara dinamis, serta menerapkan **Mitigasi Fallback CS** agar kepuasan pelanggan (CSAT) tetap terjaga saat chatbot tidak memahami pertanyaan.


### Modul 5 - Langkah 5.1: Ekstraksi Entitas Dinamis (Nomor Resi) via `re.search`
Menggunakan ekspresi reguler `r"resi:\s*(\d+)"` untuk menangkap rangkaian angka nomor resi dari dalam kalimat pertanyaan pelanggan.


```python
# ==============================================================
# MODUL 5 - LANGKAH 5.1: EKSTRAKSI ENTITAS RESI VIA REGEX
# ==============================================================

def lacak_resi_pelanggan(pesan):
    # Pola regex mencari kata 'resi:' diikuti spasi opsional dan kumpulan angka
    pola_resi = r"resi:\s*(\d+)"
    cocok = re.search(pola_resi, pesan, re.IGNORECASE)
    
    if cocok:
        nomor_resi = cocok.group(1) # Ambil grup angka hasil ekstraksi
        return {
            "ditemukan": True,
            "nomor_resi": nomor_resi,
            "status_paket": "Dalam Perjalanan ke Kota Tujuan (Manifested di DC Hub Bandung)",
            "estimasi_tiba": "Besok Siang (Maksimal Pukul 15:00 WIB)"
        }
    else:
        return {
            "ditemukan": False,
            "nomor_resi": None,
            "pesan_error": "Format nomor resi tidak terdeteksi. Silakan gunakan format: 'resi: [angka]'"
        }

# Uji coba pesan pelanggan
pesan_1 = "Tolong cek lokasi paket saya dengan nomor resi: 99482101 sekarang"
hasil_resi = lacak_resi_pelanggan(pesan_1)

print(f'Pesan Masuk: "{pesan_1}"')
if hasil_resi["ditemukan"]:
    print(f" [OK] Nomor Resi Berhasil Diekstrak : {hasil_resi['nomor_resi']}")
    print(f" [OK] Status Pelacakan Paket        : {hasil_resi['status_paket']}")
    print(f" [OK] Estimasi Tiba                 : {hasil_resi['estimasi_tiba']}")
else:
    print(f" [GAGAL] {hasil_resi['pesan_error']}")
```

### Modul 5 - Langkah 5.2: Penerapan Mitigasi Fallback CS (Handling Blok `else` via WhatsApp)
**Prinsip Bisnis:** Jika pesan pelanggan tidak dikenali oleh aturan bot (masuk ke blok `else`), sistem **dilarang keras** hanya menjawab *"Maaf saya tidak tahu"* tanpa solusi. Sistem wajib memberikan nomor tiket kendala dan tombol tautan langsung ke WhatsApp Live Agent CS agar tidak terjadi *Customer Churn*.


```python
# ==============================================================
# MODUL 5 - LANGKAH 5.2: LOGIKA MITIGASI FALLBACK ELSE CS
# ==============================================================

def handling_fallback_mitigasi_cs(pesan_pelanggan):
    # Simulasi evaluasi aturan bisnis
    is_terdeteksi = False # Contoh kasus di mana bot tidak mengenali topik pesan
    
    if is_terdeteksi:
        return "Jawaban bot otomatis..."
    else:
        # MITIGASI BISNIS CS: Jangan biarkan pelanggan kecewa tanpa jalan keluar
        no_tiket = f"TKT-{random.randint(10000, 99999)}"
        link_wa_cs = "https://wa.me/6281234567890"
        
        pesan_fallback = (
            f"Mohon maaf, pertanyaan Anda belum dapat dijawab otomatis oleh sistem bot kami.\n"
            f"Tiket Bantuan Anda: [{no_tiket}]\n"
            f"Agar keluhan/pertanyaan Anda segera ditangani oleh staf kami, silakan klik tautan WhatsApp Live Agent berikut:\n"
            f"{link_wa_cs}?text=Halo%20Admin%20FastSend,%20saya%20butuh%20bantuan%20terkait%20tiket%20{no_tiket}"
        )
        return pesan_fallback

print("Contoh Tampilan Respon Fallback saat Bot Mengalami Kendala Pemahaman:")
print("-------------------------------------------------------------------")
print(handling_fallback_mitigasi_cs("halo min mau sewa gudang kontainer 50 ton"))
```

### Modul 6 - Langkah 6.1: Pelatihan Model Naïve Bayes (`MultinomialNB`) dengan Scikit-Learn Pipeline
Menyiapkan dataset data latih (teks dan label intent), lalu melatih model statistik *Multinomial Naïve Bayes* terintegrasi dengan ekstraksi fitur TF-IDF.


```python
# ==============================================================
# MODUL 6 - LANGKAH 6.1: TRAINING PIPELINE MODEL MULTINOMIAL NAÏVE BAYES
# ==============================================================

# 1. Dataset Latih (Training Data)
X_train = [
    "halo admin fastsend", "selamat pagi min", "hai admin", "permisi kak",
    "berapa tarif ongkir ke jakarta", "biaya kirim paket ke surabaya", "cek ongkos kirim bandung",
    "tolong cek lokasi resi pengiriman", "posisi paket saya dimana min", "lacak status resi paket",
    "paket saya rusak parah", "kurir pengiriman sangat lambat", "barang pecah belah komplain",
    "alamat drop point cabang terdekat dimana", "lokasi kantor cabang fastsend bandung"
]

y_train = [
    "sapaan", "sapaan", "sapaan", "sapaan",
    "tanya_tarif", "tanya_tarif", "tanya_tarif",
    "lacak_resi", "lacak_resi", "lacak_resi",
    "komplain", "komplain", "komplain",
    "cek_lokasi", "cek_lokasi"
]

# 2. Membangun Pipeline ML: CountVectorizer -> TfidfTransformer -> MultinomialNB
model_pipeline_nb = Pipeline([
    ('vect', CountVectorizer()),
    ('tfidf', TfidfTransformer()),
    ('clf', MultinomialNB())
])

# 3. Melatih Model (Fit Training Data)
model_pipeline_nb.fit(X_train, y_train)
print("[OK] Model Machine Learning MultinomialNB Berhasil Dilatih!")

# 4. Pengujian Prediksi Intent Kalimat Baru
pesan_uji = "permisi mau tanya berapa biaya tarif kirim ke Surabaya dong"
prediksi_intent = model_pipeline_nb.predict([pesan_uji])[0]
probabilitas = max(model_pipeline_nb.predict_proba([pesan_uji])[0]) * 100

print(f'\nKalimat Uji     : "{pesan_uji}"')
print(f"Prediksi Intent : [{prediksi_intent}]")
print(f"Tingkat Keyakinan (Confidence) : {probabilitas:.2f}%")
```

```python
# ==============================================================
# MODUL 6 - LANGKAH 6.2: LOGISTIC REGRESSION & EVALUASI MODEL
# ==============================================================

# Membangun Pipeline alternatif dengan Logistic Regression
model_pipeline_lr = Pipeline([
    ('vect', CountVectorizer()),
    ('tfidf', TfidfTransformer()),
    ('clf', LogisticRegression(random_state=42))
])

# Latih model Logistic Regression
model_pipeline_lr.fit(X_train, y_train)

# Dataset Pengujian (Testing Data)
X_test = [
    "halo selamat siang",
    "berapa ongkir kirim paket",
    "cek nomor resi pengiriman",
    "kurir lambat paket rusak",
    "lokasi drop point kantor cabang"
]
y_test = ["sapaan", "tanya_tarif", "lacak_resi", "komplain", "cek_lokasi"]

y_pred_nb = model_pipeline_nb.predict(X_test)
y_pred_lr = model_pipeline_lr.predict(X_test)

print(f"Akurasi Model Naïve Bayes         : {accuracy_score(y_test, y_pred_nb) * 100:.1f}%")
print(f"Akurasi Model Logistic Regression : {accuracy_score(y_test, y_pred_lr) * 100:.1f}%")

print("\n--- Laporan Klasifikasi (Naïve Bayes) ---")
print(classification_report(y_test, y_pred_nb, zero_division=0))
```

# BAGIAN TERPADU: Sistem AI Chatbot Bisnis End-to-End (All-in-One Modul 1–6 Bersatu)
Pada bagian pamungkas ini, seluruh konsep dari **Modul 1 sampai Modul 6** disatukan menjadi sebuah kelas AI Chatbot Bisnis yang utuh (`FastSendIntegratedChatbot`).

**Alur Kerja Terpadu:**
1. **Input Demografi & Sapaan Etis (Modul 1 & 2):** Menyesuaikan diksi respon berdasarkan usia pelanggan.
2. **Preprocessing Teks Otomatis (Modul 2):** Case folding, pembersihan tanda baca, dan tokenisasi.
3. **Ekstraksi Entitas Nomor Resi (Modul 5):** Menggunakan Regex `re.search` untuk mendeteksi nomor resi jika pelanggan menanyakan paket.
4. **Prediksi Intent Machine Learning (Modul 3, 4, 6):** Model Scikit-Learn TF-IDF + MultinomialNB memprediksi maksud pertanyaan pelanggan secara cerdas.
5. **Mitigasi Fallback CS (Modul 5):** Jika nilai probabilitas/confidence rendah (< 35%) atau intent tidak dikenali, bot secara otomatis menampilkan nomor tiket bantuan dan link WhatsApp Live Agent CS.


```python
# ==============================================================
# BAGIAN TERPADU: KELAS SISTEM AI CHATBOT BISNIS LENGKAP
# ==============================================================

class FastSendIntegratedChatbot:
    def __init__(self):
        # 1. Dataset Latih Klasifikasi Intent
        self.X_train = [
            "halo admin fastsend", "selamat pagi kak", "hai admin", "permisi",
            "berapa tarif ongkir pengiriman", "biaya kirim paket ke kota lain", "cek harga ongkir",
            "tolong lacak nomor resi", "posisi paket saya dimana", "lacak status resi pengiriman",
            "paket saya rusak parah", "kurir pengiriman sangat lambat", "barang tidak sampai komplain",
            "alamat kantor cabang drop point dimana", "lokasi drop point terdekat"
        ]
        self.y_train = [
            "sapaan", "sapaan", "sapaan", "sapaan",
            "tanya_tarif", "tanya_tarif", "tanya_tarif",
            "lacak_resi", "lacak_resi", "lacak_resi",
            "komplain", "komplain", "komplain",
            "cek_lokasi", "cek_lokasi"
        ]
        
        # 2. Template Respon Bisnis Berdasarkan Intent
        self.respon_dict = {
            "sapaan": [
                "Halo! Selamat datang di FastSend Express. Ada yang bisa kami bantu terkait pengiriman Anda?",
                "Hai Kak! Senang bisa membantu Anda di FastSend Express. Silakan sampaikan pertanyaan Anda."
            ],
            "tanya_tarif": [
                "Tarif reguler kami mulai dari Rp 10.000/kg antar kota. Cek simulasi lengkap di fastsend.co.id/tarif.",
                "Biaya kirim dihitung berdasarkan jarak dan berat paket. Silakan sebutkan kota asal dan tujuan Anda."
            ],
            "lacak_resi": [
                "Paket Anda dengan nomor resi terdaftar sedang dalam perjalanan menuju kota tujuan.",
                "Status terkini: Paket telah lolos sortir di Hub Transit Pusat dan siap diantar kurir."
            ],
            "komplain": [
                "Mohon maaf atas kendala yang dialami. Laporan Anda telah masuk ke tim kendali mutu kami untuk diproses.",
                "Kami memohon maaf atas ketidaknyamanan Anda. Klaim garansi paket rusak/terlambat akan diproses 1x24 jam."
            ],
            "cek_lokasi": [
                "Kantor cabang dan drop point FastSend buka setiap hari pukul 08:00 - 21:00 WIB di seluruh kota besar.",
                "Drop point terdekat tersedia di radius 2 km dari pusat kota. Silakan cek map di aplikasi FastSend."
            ]
        }
        
        # 3. Latih Model ML Pipeline (CountVectorizer + Tfidf + MultinomialNB)
        self.model = Pipeline([
            ('vect', CountVectorizer()),
            ('tfidf', TfidfTransformer()),
            ('clf', MultinomialNB())
        ])
        self.model.fit(self.X_train, self.y_train)
        
    def _get_sapaan_sesuai_umur(self, nama, umur):
        if umur < 18:
            return f"Halo Sobat {nama}"
        elif 18 <= umur <= 50:
            return f"Halo Kak {nama}"
        else:
            return f"Selamat Pagi/Siang Bapak/Ibu {nama}"
            
    def proses_pesan(self, pesan, user_info):
        nama = user_info.get("nama", "Pelanggan")
        umur = user_info.get("umur", 25)
        sapaan = self._get_sapaan_sesuai_umur(nama, umur)
        
        # 1. Preprocessing Sederhana (Case Folding & Clean)
        pesan_clean = re.sub(r'[^a-zA-Z0-9\s:]', ' ', pesan.lower()).strip()
        
        # 2. Ekstraksi Entitas Regex: Cek Nomor Resi
        cocok_resi = re.search(r"resi:\s*(\d+)", pesan, re.IGNORECASE)
        nomor_resi = cocok_resi.group(1) if cocok_resi else None
        
        # 3. Prediksi Intent via Machine Learning Pipeline
        prediksi_intent = self.model.predict([pesan_clean])[0]
        probabilitas = max(self.model.predict_proba([pesan_clean])[0]) * 100
        
        # 4. Mitigasi Fallback CS jika Confidence < 30%
        if probabilitas < 30.0:
            no_tiket = f"TKT-{random.randint(10000, 99999)}"
            link_wa = "https://wa.me/6281234567890"
            respon = (
                f"{sapaan}, mohon maaf kami belum dapat mengenali permintaan Anda secara spesifik.\n"
                f"Nomor Tiket Antrean: [{no_tiket}]\n"
                f"Hubungi WhatsApp Live Agent CS kami: {link_wa}?text=Bantuan%20Tiket%20{no_tiket}"
            )
            return {"intent": "fallback", "confidence": probabilitas, "respon": respon}
            
        # 5. Bangun Jawaban Bisnis Berdasarkan Intent
        jawaban_dasar = random.choice(self.respon_dict.get(prediksi_intent, ["Ada yang bisa dibantu?"]))
        if nomor_resi and prediksi_intent == "lacak_resi":
            jawaban_lengkap = f"{sapaan}! {jawaban_dasar}\n[INFO RESI {nomor_resi}]: Estimasi tiba besok sore pukul 15:00 WIB."
        else:
            jawaban_lengkap = f"{sapaan}! {jawaban_dasar}"
            
        return {
            "intent": prediksi_intent,
            "confidence": probabilitas,
            "nomor_resi": nomor_resi,
            "respon": jawaban_lengkap
        }

# Inisialisasi Chatbot Terpadu
chatbot_fastsend = FastSendIntegratedChatbot()
print("FastSend Integrated AI Chatbot Siap Melayani Pengujian!")
```

## SOAL 4: Ekstraksi Entitas Nomor Resep Medis via Regex

**Instruksi Soal:**
Buat fungsi `verifikasi_resep(pesan)` menggunakan `re.search` dengan pola regex `resep\s*:\s*(\d+)` untuk mengambil nomor resep medis dari kalimat masukan: `'Tolong cek lokasi racikan resep: 88720199'`.


```python
# SOAL 4: REGEX EKSTRAKSI NOMOR RESEP MEDIS
import re

def verifikasi_resep(pesan):
    match = re.search(r'resep\s*:\s*(\d+)', pesan, re.IGNORECASE)
    if match:
        no_resep = match.group(1)
        return f'Status Resep Medis #{no_resep} sedang diracik oleh apoteker.'
    return 'Nomor resep tidak terdeteksi. Format resep: resep: [Nomor]'

# Pengujian Ekstraksi Entitas
pesan_resep = 'Tolong cek lokasi racikan resep: 88720199'
print('Pesan Pasien :', pesan_resep)
print('Hasil Verifikasi:', verifikasi_resep(pesan_resep))
```

```python
# SOAL 5: ML INTENT CLASSIFICATION PIPELINE
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import CountVectorizer, TfidfTransformer
from sklearn.naive_bayes import MultinomialNB

# Data Latih Intent Farmasi
X_train_pharma = [
    'halo admin',
    'minta info harga vitamin',
    'jadwal praktek dokter kapan',
    'obat saya belum sampai'
]
y_train_pharma = [
    'sapaan',
    'tanya_harga',
    'tanya_dokter',
    'komplain'
]

# Pipeline Pembelajaran Mesin
model_pharma = Pipeline([
    ('vect', CountVectorizer()),
    ('tfidf', TfidfTransformer()),
    ('clf', MultinomialNB())
])

# Melatih Model
model_pharma.fit(X_train_pharma, y_train_pharma)

# Prediksi Kalimat Uji
kalimat_uji = 'permisi mau tanya berapa harga obat paracetamol dong'
prediksi_intent = model_pharma.predict([kalimat_uji])[0]

print('Kalimat Uji    :', kalimat_uji)
print('Prediksi Intent:', prediksi_intent)
```

## SOAL 4: Ekstraksi Entitas Nomor Resi Pengiriman via Regex

**Instruksi Soal:**
Buat fungsi `lacak_resi(pesan)` menggunakan `re.search` dengan pola regex `resi\s*:\s*(\d+)` untuk mengekstrak nomor resi berformat angka dari kalimat masukan: `'Tolong cek lokasi paket resi: 99482101'`.


```python
# SOAL 4: REGEX EKSTRAKSI NOMOR RESI PENGIRIMAN
import re

def lacak_resi(pesan):
    match = re.search(r'resi\s*:\s*(\d+)', pesan, re.IGNORECASE)
    if match:
        no_resi = match.group(1)
        return f'Status Resi Paket #{no_resi} sedang dalam perjalanan pengiriman kurir FastSend.'
    return 'Nomor resi tidak terdeteksi. Format resi: resi: [Nomor]'

# Pengujian Ekstraksi Entitas
pesan_resi = 'Tolong cek lokasi paket resi: 99482101'
print('Pesan Pelanggan :', pesan_resi)
print('Hasil Pelacakan:', lacak_resi(pesan_resi))
```

## SOAL 5: Pelatihan & Prediksi Model ML Intent Classification

**Instruksi Soal:**
Gunakan Scikit-Learn `Pipeline` (`CountVectorizer` + `TfidfTransformer` + `MultinomialNB`) untuk melatih model klasifikasi intent FastSend.
* Data latih X: `['halo admin', 'minta info tarif ongkir', 'lokasi drop point dimana', 'paket saya rusak']`
* Label target y: `['sapaan', 'tanya_tarif', 'cek_lokasi', 'komplain']`
* Uji prediksi intent pada kalimat baru: `'permisi mau tanya berapa biaya tarif kirim ke Surabaya dong'`.


```python
# SOAL 5: ML INTENT CLASSIFICATION PIPELINE
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import CountVectorizer, TfidfTransformer
from sklearn.naive_bayes import MultinomialNB

# Data Latih Intent Ekspedisi FastSend
X_train_fastsend = [
    'halo admin',
    'minta info tarif ongkir',
    'lokasi drop point dimana',
    'paket saya rusak'
]
y_train_fastsend = [
    'sapaan',
    'tanya_tarif',
    'cek_lokasi',
    'komplain'
]

# Pipeline Pembelajaran Mesin
model_fastsend = Pipeline([
    ('vect', CountVectorizer()),
    ('tfidf', TfidfTransformer()),
    ('clf', MultinomialNB())
])

# Melatih Model
model_fastsend.fit(X_train_fastsend, y_train_fastsend)

# Prediksi Kalimat Uji
kalimat_uji = 'permisi mau tanya berapa biaya tarif kirim ke Surabaya dong'
prediksi_intent = model_fastsend.predict([kalimat_uji])[0]

print('Kalimat Uji    :', kalimat_uji)
print('Prediksi Intent:', prediksi_intent)
```
