## SOAL 1: Input Demografi Pasien, Percabangan Layanan & Diksi Usia

**Instruksi Soal:**
1. **(A)** Buat masukan data untuk 3 parameter: Nama Pasien, Nomor HP, dan Umur Pasien.
2. **(B)** Terapkan logika percabangan untuk personalisasi sapaan dan paket layanan kesehatan:
   * Umur < 18 tahun: Sapaan 'Halo Kak', Layanan 'Package Vitamin Pelajar & Booster Imunitas Multivitamin'.
   * Umur 18 - 50 tahun: Sapaan 'Halo Kak', Layanan 'Voucher Telekonsultasi Dokter & Vitamin Work-Life Balance'.
   * Umur > 50 tahun: Sapaan 'Selamat Pagi/Siang Bapak/Ibu', Layanan 'Paket Cek Rutin Kolesterol/Gula Darah & Free Home Delivery Obat'.
3. **(C)** Berikan penjelasan analisis pada sel teks terkait etika diksi dan relevansi layanan lansia.


```python
# SOAL 1: INPUT DEMOGRAFI PASIEN, DIKSI & LAYANAN KESEHATAN
nama_pasien = input('Masukkan Nama Pasien: ') or 'Budi Santoso'
hp_pasien = input('Masukkan No HP: ') or '08123456789'
umur_input = input('Masukkan Umur Pasien: ')
umur_pasien = int(umur_input) if umur_input.strip() else 55

if umur_pasien < 18:
    sapaan = f'Halo Kak {nama_pasien}'
    layanan = 'Package Vitamin Pelajar & Booster Imunitas Multivitamin'
elif 18 <= umur_pasien <= 50:
    sapaan = f'Halo Kak {nama_pasien}'
    layanan = 'Voucher Telekonsultasi Dokter & Vitamin Work-Life Balance'
else:
    sapaan = f'Selamat Pagi/Siang Bapak/Ibu {nama_pasien}'
    layanan = 'Paket Cek Rutin Kolesterol/Gula Darah & Free Home Delivery Obat'

print(f'\nRespon CS: "{sapaan}! Selamat datang di Smart-Pharma CS."')
print(f'Penawaran Layanan Khusus: {layanan}')
```

## SOAL 2: Preprocessing Data Teks Keluhan Layanan Apotek

**Instruksi Soal:**
Diberikan teks ulasan mentah: `'Pelayanan Obat SANGAT Lambat!! Resep Dokter Saya Belum Dikirim dan Dus Pecah :-('`.
Terapkan tahapan NLP:
1. Case Folding (konversi ke huruf kecil).
2. Pembersihan simbol dan tanda baca via Regex.
3. Tokenisasi kalimat menjadi kata via NLTK `word_tokenize`.
4. Stopword Removal dan Stemming Bahasa Indonesia via Sastrawi.


```python
# SOAL 2: PREPROCESSING DATA TEKS KELUHAN PASIEN
# Hilangkan tanda komentar jika instalasi diperlukan di Google Colab
try:
    import Sastrawi
except ImportError:
    import sys, subprocess
    subprocess.check_call([sys.executable, "-m", "pip", "install", "Sastrawi", "-q"])

import re
try:
    import nltk
    nltk.download('punkt', quiet=True)
    from nltk.tokenize import word_tokenize
except Exception:
    def word_tokenize(text):
        return re.findall(r'\w+|[^\w\s]', text)

from Sastrawi.StopWordRemover.StopWordRemoverFactory import StopWordRemoverFactory
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory

teks_mentah = 'Pelayanan Obat SANGAT Lambat!! Resep Dokter Saya Belum Dikirim dan Dus Pecah :-('

# 1. Case Folding & Pembersihan Karakter Khusus / Simbol via Regex
teks_clean = re.sub(r'[^a-zA-Z0-9\s]', '', teks_mentah.lower())

# 2. Tokenisasi Kalimat
tokens = word_tokenize(teks_clean)

# 3. Stopword Removal & Stemming Bahasa Indonesia (Sastrawi)
stop_remover = StopWordRemoverFactory().create_stop_word_remover()
stemmer = StemmerFactory().create_stemmer()
teks_stemmed = stemmer.stem(stop_remover.remove(teks_clean))

print('Teks Clean :', teks_clean)
print('Tokens     :', tokens)
print('Stemming   :', teks_stemmed)
```

## SOAL 3: Handling Fallback Else dengan Mitigasi Bisnis CS Apoteker

**Instruksi Soal:**
1. **(A)** Buat fungsi `chatbot_pharma(pesan)`. Jika pesan masukan tidak cocok dengan intent dasar (masuk blok `else`), bot dilarang hanya menjawab 'tidak mengerti'. Terapkan mitigasi bisnis berupa eskalasi ke link WhatsApp Live Agent Apoteker Duty (`https://wa.me/6281199001122`).
2. **(B)** Jelaskan dampak bisnis dan keselamatan pasien pada sel teks.


```python
# SOAL 3: FALLBACK ELSE DENGAN MITIGASI BISNIS CS APOTEKER
def chatbot_pharma(pesan):
    pesan_clean = pesan.lower().strip()
    if 'harga' in pesan_clean or 'vitamin' in pesan_clean:
        return 'Katalog harga vitamin dan suplemen tersedia lengkap di aplikasi Smart-Pharma.'
    elif 'dokter' in pesan_clean or 'jadwal' in pesan_clean:
        return 'Jadwal telekonsultasi dokter umum buka setiap hari pukul 08.00 - 20.00 WIB.'
    else:
        # Mitigasi Bisnis CS Kesehatan (Safety & Patient Support)
        return (
            'Chatbot: Mohon maaf Kak, pertanyaan Anda membutuhkan penanganan khusus CS Apoteker.\n' +
            'Hubungi Live Agent CS via WhatsApp: https://wa.me/6281199001122 (Fast Response).'
        )

# Pengujian Pesan Diluar Intent Dasar
pesan_uji = 'Apakah obat ini aman diminum bersamaan dengan kopi?'
print('Input Pesan :', pesan_uji)
print('Respon CS   :')
print(chatbot_pharma(pesan_uji))
```

### Penjelasan Sel Teks Soal 3:
1. **Keselamatan Pasien (Patient Safety):**
   Di sektor farmasi dan layanan medis, kesalahan jawaban otomatis oleh sistem chatbot dapat berakibat fatal pada interaksi obat dan kesehatan pasien. Kegagalan memahami pertanyaan obat harus segera dialihkan ke apoteker resmi.
2. **Mencegah Customer Churn & Menjaga Skor CSAT:**
   Mengalihkan pengguna langsung ke WhatsApp petugas apoteker berlisensi menjamin pasien mendapatkan jawaban yang valid dan akurat secara cepat. Hal ini memelihara *Customer Satisfaction Score* (CSAT) dan mencegah pasien berpindah ke platform telemedicine lain (*customer churn*).


## SOAL 4: Ekstraksi Entitas Nomor Resep Medis via Regex

**Instruksi Soal:**
Buat fungsi `verifikasi_resep(pesan)` menggunakan `re.search` dengan pola regex `resep\s*:\s*(\d+)` untuk mengambil nomor resep medis dari kalimat masukan: `'Tolong cek lokasi racikan resep: 88720199'`.


```python
# SOAL 4: REGEX EKSTRAKSI NOMOR RESEP MEDIS
import re

def verifikasi_resep(pesan):
    match = re.search(r'resep\s*:\s*(\d+)', pesan, re.IGNORECASE)
    if match:
        no_resep = match.group(1)
        return f'Status Resep Medis #{no_resep} sedang diracik oleh apoteker.'
    return 'Nomor resep tidak terdeteksi. Format resep: resep: [Nomor]'

# Pengujian Ekstraksi Entitas
pesan_resep = 'Tolong cek lokasi racikan resep: 88720199'
print('Pesan Pasien :', pesan_resep)
print('Hasil Verifikasi:', verifikasi_resep(pesan_resep))
```

## SOAL 5: Pelatihan & Prediksi Model ML Intent Classification

**Instruksi Soal:**
Gunakan Scikit-Learn `Pipeline` (`CountVectorizer` + `TfidfTransformer` + `MultinomialNB`) untuk melatih model klasifikasi intent Smart-Pharma.
* Data latih X: `['halo admin', 'minta info harga vitamin', 'jadwal praktek dokter kapan', 'obat saya belum sampai']`
* Label target y: `['sapaan', 'tanya_harga', 'tanya_dokter', 'komplain']`
* Uji prediksi intent pada kalimat baru: `'permisi mau tanya berapa harga obat paracetamol dong'`.


```python
# SOAL 5: ML INTENT CLASSIFICATION PIPELINE
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import CountVectorizer, TfidfTransformer
from sklearn.naive_bayes import MultinomialNB

# Data Latih Intent Farmasi
X_train_pharma = [
    'halo admin',
    'minta info harga vitamin',
    'jadwal praktek dokter kapan',
    'obat saya belum sampai'
]
y_train_pharma = [
    'sapaan',
    'tanya_harga',
    'tanya_dokter',
    'komplain'
]

# Pipeline Pembelajaran Mesin
model_pharma = Pipeline([
    ('vect', CountVectorizer()),
    ('tfidf', TfidfTransformer()),
    ('clf', MultinomialNB())
])

# Melatih Model
model_pharma.fit(X_train_pharma, y_train_pharma)

# Prediksi Kalimat Uji
kalimat_uji = 'permisi mau tanya berapa harga obat paracetamol dong'
prediksi_intent = model_pharma.predict([kalimat_uji])[0]

print('Kalimat Uji    :', kalimat_uji)
print('Prediksi Intent:', prediksi_intent)
```
